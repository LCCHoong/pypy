```python
traindf['Gender'].unique()
```

```python
traindf['Gender'] = traindf['Gender'].fillna('Missing')
```

```python
traindf['Gender'] = traindf['Gender'].astype('category')
```

```python
traindf.isnull().sum()
```

```python
traindf.dtypes
#married -cat , dependents -cat  , education - bool ? , self employed -bool ?
```

```python
traindf['Married'] = traindf['Married'].fillna('Missing')
traindf['Married'] = traindf['Married'].astype('category')
```

```python
traindf.isnull().sum()
```

```python
traindf['Dependents'] = traindf['Dependents'].fillna('Missing')
traindf['Dependents'] = traindf['Dependents'].astype('category')
```

```python
traindf.isnull().sum()
```

```python
traindf['Education'] = traindf['Dependents'].fillna(traindf['Dependents'].mode()[0])
```

```python
traindf['Education'] = traindf['Education'].astype('bool')
```

```python
traindf.isnull().sum()
```

```python
traindf['Self_Employed'] = traindf['Self_Employed'].fillna('Missing')
traindf['Self_Employed'] = traindf['Self_Employed'].astype('category')
```

```python
traindf.isnull().sum()
```
