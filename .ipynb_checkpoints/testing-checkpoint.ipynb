```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
traindf = pd.read_csv("training_set.csv")
```

```python
traindf.describe()
```

```python
traindf.dtypes
```

```python
traindf
```

```python
traindf.drop_duplicates(inplace=True)
```

```python
traindf
```

```python
traindf.drop('Loan_ID', axis = 1 , inplace =True)
```

```python
traindf
```

```python
traindf.isnull().sum()
```

```python
traindf['Gender'].unique()
```

```python
traindf['Gender'] = traindf['Gender'].fillna(traindf['Gender'].mode()[0])
```

```python
traindf['Gender'] = traindf['Gender'].astype('category')
```

```python
traindf.isnull().sum()
```

```python
traindf.dtypes
#married -cat , dependents -cat  , education - bool ? , self employed -bool ?
```

```python
traindf['Married'] = traindf['Married'].fillna(traindf['Married'].mode()[0])
traindf['Married'] = traindf['Married'].astype('category')
```

```python
traindf.isnull().sum()
```

```python
traindf['Dependents'] = traindf['Dependents'].fillna(traindf['Dependents'].mode()[0])
traindf['Dependents'] = traindf['Dependents'].astype('category')
```

```python
traindf.isnull().sum()
```

```python
traindf['Education'] = traindf['Education'].fillna(traindf['Dependents'].mode()[0])
```

```python
traindf['Education'] = traindf['Education'].astype('bool')
```

```python
traindf.isnull().sum()
```

```python
traindf['Self_Employed'] = traindf['Self_Employed'].fillna('Missing')
traindf['Self_Employed'] = traindf['Self_Employed'].astype('category')
```

```python
traindf.isnull().sum()
```

```python
traindf.dtypes
```

```python
traindf['ApplicantIncome'].skew()
```

```python
traindf['ApplicantIncome'] = traindf['ApplicantIncome'].fillna(traindf['ApplicantIncome'].median())
```

```python
traindf.isnull().sum()
```

```python
traindf['CoapplicantIncome'].skew()
```

```python
traindf['CoapplicantIncome'] = traindf['CoapplicantIncome'].fillna(traindf['CoapplicantIncome'].median())
```

```python
traindf.isnull().sum()
```

```python
traindf['LoanAmount'].skew()
```

```python
traindf['LoanAmount'] = traindf['LoanAmount'].fillna(traindf['LoanAmount'].median())
```

```python
traindf.isnull().sum()
```

```python
traindf['Loan_Amount_Term'].unique()
```

```python
traindf['Loan_Amount_Term'] = traindf['Loan_Amount_Term'].fillna(traindf['Loan_Amount_Term'].mode()[0])
traindf['Loan_Amount_Term'] = traindf['Loan_Amount_Term'].astype('category')
```

```python
traindf.isnull().sum()
```

```python
traindf['Credit_History'].unique()
```

```python
traindf['Credit_History'] = traindf['Credit_History'].fillna(traindf['Credit_History'].mode()[0])
traindf['Credit_History'] = traindf['Credit_History'].astype('bool')
```

```python
traindf.isnull().sum()
```

```python
traindf.drop_duplicates(inplace=True)
```

```python
traindf
```

```python
traindf.dtypes
```
