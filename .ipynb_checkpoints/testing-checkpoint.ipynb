```python
traindf['Gender'].unique()
```

```python
traindf['Gender'] = pd.Categorical(traindf['Gender'], categories = ['Male' , 'Female'])
```

```python
traindf['Gender'] = traindf['Gender'].fillna('Missing') 
```

```python
traindf.isnull().sum()
```

```python
traindf.dtypes
```

```python
traindf['Married'] = traindf['Married'].map({'Yes' : 1 , 'No' : 0})
```

```python
traindf['Married'] = traindf['Married'].fillna('Missing')
```

```python
traindf.isnull().sum()
```

```python
traindf['Dependents'].unique()
```

```python
traindf['Dependents'] = traindf['Dependents'].astype('string')
```

```python
traindf['Dependents'] = pd.Categorical(traindf['Dependents'], categories = ['0','1','2','3+'])
```

```python
traindf['Dependents'] = traindf['Dependents'].fillna('Missing')
```
